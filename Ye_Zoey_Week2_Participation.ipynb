```python
letters = ['a', 'b', 'c', 'd']

print(letters[int(int('3' * 2) // 11)])
#'3' * 2 becomes '33'
#int('33') becomes 33
#33 // 11 = 3
#So letters[3] evaluates to 'd'

print(letters[-1])
#-1 refers to the last item in a list so letters[-1] evaluates to 'd'


print(letters[:2])
#[:2] returns elements from the beginning of the list up to, but not including, index 2. Therefore, letters[:2] evaluates to ['a', 'b'].
```

```python
spam = ['apples', 'bananas', 'tofu', 'cats']

def comma_code(items):
    if len(items) == 0:
        return ''
    elif len(items) == 1:
        return str(items[0])
    else:
        return ', '.join(str(item) for item in items[:-1]) + ', and ' + str(items[-1])

print(comma_code(spam))
```
